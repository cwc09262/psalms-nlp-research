```python
import pandas as pd

psalms = pd.read_csv("cleaned_psalm_verses.csv")
psalms
```

```python
def word_count(text): 
    # splitting based on whitesapce
    words = text.split()
    # returning the number of words
    return len(words)
```

```python
psalms['word_count'] = psalms['verse'].apply(word_count)

psalms
```

```python
# Calculate character count for each verse
psalms['char_count'] = psalms['verse'].apply(len)
psalms
```

```python
psalms.columns
```

## Per Psalm

### Average Number of Verses

```python

# Calculate total word count for each Psalm
total_word_count_per_psalm = psalms.groupby('text')['word_count'].sum().reset_index()

# Calculate the average word count for all Psalms
average_word_count = total_word_count_per_psalm['word_count'].mean()

# Display the results
print(total_word_count_per_psalm)
print(f"Average word count for all Psalms: {average_word_count:.2f}")
```

### Psalm Word Count per Psalm

```python

# Calculate the average word count per Psalm
average_word_count_per_psalm = psalms.groupby(['text', 'psalm_num'])['word_count'].sum().reset_index()
average_word_count_per_psalm
```

```python
average_word_count_per_psalm['word_count'].mean()
```

```python

# Calculate total word count for each Psalm
average_word_count_per_psalm = psalms.groupby(['text', 'psalm_num'])['word_count'].sum().reset_index()

# Calculate the average word count across all Psalms
overall_average_word_count = average_word_count_per_psalm['word_count'].mean()

# Display the DataFrame and the average
print(average_word_count_per_psalm)
print(f"Overall Average Word Count per Psalm: {overall_average_word_count:.0f}")
```

### Psalm Character Count 
#### Per Psalm

```python

# Calculate the average word count per Psalm
average_word_count_per_psalm = psalms.groupby(['text', 'psalm_num'])['char_count'].sum().reset_index()
average_word_count_per_psalm
```

```python
average_word_count_per_psalm['word_count'].mean()
```

```python

# Calculate total word count for each Psalm
average_word_count_per_psalm = psalms.groupby(['text', 'psalm_num'])['char_count'].sum().reset_index()

# Calculate the average word count across all Psalms
overall_average_char_count = average_word_count_per_psalm['char_count'].mean()

# Display the DataFrame and the average
print(average_word_count_per_psalm)
print(f"Overall Average Character Count per Psalm: {overall_average_char_count:.0f}")
```

#### Per Verse

```python

# Calculate total character count for each verse
total_char_count_per_verse = psalms.groupby(['text', 'psalm_num', 'verse_num'])['char_count'].sum().reset_index()

# Calculate the average character count across all verses
overall_average_char_count = total_char_count_per_verse['char_count'].mean()

# Display the DataFrame and the average
print(total_char_count_per_verse)
print(f"Overall Average Character Count per Verse: {overall_average_char_count:.0f}")
```

## Average Word Count Per Verse

```python

# Calculate total word count for each Psalm
total_word_count_per_psalm = psalms.groupby('text')['word_count'].sum().reset_index()

# Calculate the average word count for all Psalms
average_word_count = total_word_count_per_psalm['word_count'].mean()

# Display the results
print(total_word_count_per_psalm)
print(f"Average word count for all Psalms: {average_word_count:.2f}")
```

### Average Word Count

```python
# Create a pivot table to find the average word count and character count per verse
pivot_table = pd.pivot_table(
    psalms,
    index=['text', 'psalm_num', 'verse_num'],  # Grouping by text, psalm number, and verse number
    values=['word_count', 'char_count'],        # Values we want to average
    aggfunc='mean'                              # Using mean to get the average
)
# Calculate averages
average_word_count = pivot_table['word_count'].mean()
average_char_count = pivot_table['char_count'].mean()

# Print formatted output
print(f"Average Word Count: {average_word_count:.0f}")
print(f"Average Character Count: {average_char_count:.0f}")
```
